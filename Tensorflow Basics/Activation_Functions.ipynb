```python
# Import matplotlib, numpy and math
import matplotlib.pyplot as plt
import numpy as np
import math

x = np.linspace(-10, 10, 100)
z = 1/(1 + np.exp(-x))

plt.plot(x, z)
plt.xlabel("x")
plt.ylabel("Sigmoid(X)")

plt.show()
```

```python
# Import matplotlib, numpy and math
import matplotlib.pyplot as plt
import numpy as np
import math

x = np.linspace(-10, 10, 100)
z = (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

plt.plot(x, z)
plt.xlabel("x")
plt.ylabel("Tanh(X)")

plt.show()
```

```python
# Import matplotlib, numpy and math
import matplotlib.pyplot as plt
import numpy as np
import math

x = np.linspace(-10, 10, 100)
z = np.maximum(0,x)

plt.plot(x, z)
plt.xlabel("x")
plt.ylabel("ReLU(X)")

plt.show()
```

```python
# Import matplotlib, numpy and math
import matplotlib.pyplot as plt
import numpy as np
import math

x = np.linspace(-10, 10, 100)
z = np.maximum(0.1*x,x)

plt.plot(x, z)
plt.xlabel("x")
plt.ylabel("Leaky ReLU(X)")

plt.show()
```

```python
import numpy as np
#https://medium.com/@amit25173/understanding-softmax-with-numpy-b7273d8ab205
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # Stabilizing to prevent overflow
    return exp_x / np.sum(exp_x)

scores = [2.0, 1.0, 0.1]
probabilities = softmax(scores)
print("Softmax Probabilities:", probabilities)
# Output: [0.65900114 0.24243297 0.09856589
```
